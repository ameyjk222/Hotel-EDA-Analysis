```python
df["reservation_status_date"] = pd.to_datetime(df["reservation_status_date"])
```

```python
hotel_counts = df["hotel"].value_counts()
print("hotel_counts")
```

```python
ax = hotel_counts.plot(kind='bar', color=["Skyblue", "Orange"])
plt.title("Booking by hotel type")
plt.xlabel("Hotel type")
plt.ylabel("number of Bookings")
plt.bar_label(ax.containers[0], fmt="%.0f", padding=3)
plt.show()
```

```python
segment_counts = df["market_segment"].value_counts()
print("segment_counts")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


plt.figure(figsize=(8,5))
ax = sns.barplot(x=segment_counts.index, y=segment_counts.values,
    palette=['lightgreen', 'orange', 'yellow', 'skyblue', 'red', 'purple']
    )

plt.title("Booking by market segment")
plt.xlabel("Market segment")
plt.ylabel("Number of bookings")
ax.bar_label(ax.containers[0], fmt="%.0f", padding=3)
plt.show()
plt.savefig("booking_by_market_segment.png")
```

```python
correlation = df["lead_time"].corr(df["is_canceled"])
print("correlation")
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x = "lead_time",
                y="is_canceled", data=df, color = 'green')
plt.title("lead_time by booking cancellation")
plt.xlabel("lead_time by booking cancellation")
plt.ylabel("is_canceled")
plt.show()
plt.savefig("lead_time_by_booking_cancellation.png")
```

```python
### what is the average lead time for bookings ?

average_lead_time = df["lead_time"].mean()
print("average_lead_time:",average_lead_time)
```

```python
### What is the distribution of bookings by hotel types ?

hotel_counts = df["hotel"].value_counts()

print("hotel_counts:", hotel_counts)

plt.figure(figsize = (8,5))
sns.barplot (x=hotel_counts.index, y=hotel_counts.values, palette = ['darkblue','darkgreen'])
plt.title ("Booking by hotel type")
plt.xlabel("hotel type")
plt.ylabel ("number of bookings")
plt.show()
plt.savefig("booking_by_hotel_type.png")
```

```python
### How many bookings were cancelled ?

booking_canceled = df["is_canceled"].sum()
print("booking_canceled:", booking_canceled)
```

```python
### what is the most common arrival month for bookings ?
common_month = df["arrival_date_month"].mode()
print("common_month:", common_month)
```

```python
### What is the average number of special requests per booking ?

special_request_per_booking =df ["total_of_special_requests"].mean()
print("special_request_per_booking:", special_request_per_booking)
```

```python
### Which country has the highest number of bookings ?

country_booking = df ["country"].value_counts()
print("country_booking:", country_booking)

country_booking = df ["country"].value_counts().max()
print("country_booking:", country_booking)
```

```python
### What is the average daily rate (ADR) for each hotel type ?

ADR_per_hotel = df.groupby("hotel")['adr'].mean()

print("ADR_per_hotel:", ADR_per_hotel)
```

```python
### What percentage of guests required car parking spaces ?
guest_parking = df["required_car_parking_spaces"].sum()
print("guest_parking:", guest_parking)

parking_percentage = (guest_parking/len(df)) * 100
print(f"parking_percentage: {parking_percentage:.2f}%")
```

```python
### What is the average stay duration in week nights and weekend nights ?
stay_duration = df["stays_in_week_nights"] + df["stays_in_weekend_nights"]
print("stay_duration:", stay_duration)

avg_stay_duration = stay_duration.mean()
print("avg_stay_duration:", avg_stay_duration)
```

```python
### How many bookings were made through travel agent ?
travel_agent = df["agent"].value_counts()
print("travel_agent:", travel_agent)
```

```python
### What is the cancellation rate for each hotel type ?
cancellation_rate = df.groupby("hotel")["is_canceled"].mean()
print("cancellation_rate:", cancellation_rate)

cancellation_percentage = cancellation_rate * 100
print("cancellation_percentage:", cancellation_percentage)
```

```python
### What is the average ADR per market segment ?
ADR_per_market_segement = df.groupby("market_segment")["adr"].mean()
print("ADR_per_market_segement:", ADR_per_market_segement)
```

```python
### What is the relationship between lead time and cancellation rate ?
relationship = df.groupby("lead_time")["is_canceled"].mean()
print("relationship:", relationship)


plt.figure(figsize=(8,4))
sns.scatterplot (x="lead_time", y="is_canceled", data=df, alpha = 0.3)
plt.title("Relation between lead time and cancellation rate")
plt.xlabel("lead_time(Days)")
plt.ylabel("is_canceled(0=No, 1=Yes)")
plt.show()
```

```python
### Which distribution channel has the highest number of booking ?
distribution_channel = df["distribution_channel"].value_counts()

print("distribution_channel:", distribution_channel)
```

```python
### What is the average number of previous cancellations by hotel type ?
cancellation_by_hotel = df.groupby ("hotel")["previous_cancellations"].mean()
print("cancellation_by_hotel:", cancellation_by_hotel)
```

```python
### What is the trend of ADR over the years ?

adr_trend = df.groupby ("adr")["arrival_date_year"].mean()
print("adr_trend:", adr_trend)

sns.lineplot( x="arrival_date_year", y="adr", data=df)
plt.title("Trend of ADR over the years")
plt.xlabel("arrival_date_year")
plt.ylabel("adr")
plt.show()
```

```python
### Which month has the highest revenue ?
revenue_by_month = df.groupby("arrival_date_month")["adr"].sum()
print("revenue_by_month:", revenue_by_month)
```

```python
### What is the impact of special requests on ADR ?
adr_special = df.groupby("total_of_special_requests")["adr"].mean()
print("adr_special:", adr_special)
```

```python
### What is the average stay duration for repeated guests versus new guests ?
total_stay_duration
df["total_stay_duration"] = df["stays_in_week_nights"] + df["stays_in_weekend_nights"]

avg_stay = df.groupby("is_repeated_guest")["total_stay_duration"].mean()
print("avg_stay:", avg_stay)

avg_stay.index = ["New Guests", "Repeated Guests"]
print("avg_stay:", avg_stay)
```

```python
### Which room type has the highest number of bookings ?

room_type = df["reserved_room_type"].value_counts()
print("room_type:", room_type)

room_type.index = ["Room Type A", "Room Type B", "Room Type C", "Room Type D", "Room Type E", "Room Type F", "Room Type G", "Room Type H", "Room Type I"]
print("room_type:", room_type)
```
